 Показываем уникальные значения для (потенциально) категориальных признаков

```python
from prettytable import PrettyTable

uniques_tb = PrettyTable()
uniques_tb.field_names = ['feature', 'n_unique', 'Values (if quantity less than 40)']
category_like_columns = []

for col in df_pl.columns:
    unique_count = df_pl[col].n_unique()
    if unique_count < 40:
        category_like_columns.append(col)
        unique_values = df_pl[col].drop_nulls().unique().sort().to_list()
        values_str = ', '.join(map(str, unique_values))
    else:
        values_str = '-'

    uniques_tb.add_row([col, unique_count, values_str])

print(uniques_tb)
```

Теперь выведем таблицы для каждого (потенциально) категориального признака и посмотрим распределение сэмплов по каждому из значений. Дальше при обработке я буду руководствоваться похожими агрегатами, чтобы удалять из сэмплов нерепрезентативные классы.

```python
from prettytable import PrettyTable

def print_unique_values_table(df, category):
    print(f'\n{"="*50}')
    print(f'Unique values in {category}:')
    print(f'\n{"="*50}')
    category_counts = df.group_by(category).agg(pl.count().alias('count')).sort('count', descending=True)
    table = PrettyTable()
    table.field_names = [category, 'count', 'percent (%)']

    total_count = category_counts['count'].sum()
    for row in category_counts.iter_rows(named=True):
        value = row[category]
        count = row['count']
        percentage = (count / total_count) * 100
        table.add_row([value, count, f'{percentage:.1f}%'])

    print(table)

for category in category_like_columns:
    print_unique_values_table(df_pl, category)
```

Из этих таблиц видно, что:

1) для `rooms` можно оставить только те сэмплы, где количество комнат не больше 6, так как в целом очень большие доли в датасете у однокомнатных и двухкомнатных квартир
2) для `floors_total` и `floor` можно не беспокоиться об удалении маломощных классов (меньше 5% от всего датасета)
3) `studio` - очень несбалансированная булева фича (можно в целом отказаться от ее использования, и опираться только на количество комнат. Для студии - 0 комнат)

### 2.2 Обработка пропущенных значений


Вновь посмотрим, сколько пропущенных значений в каждом столбце:

```python
def analyze_missing_values_by_column(df):
    total_rows = df.height

    missing_analysis = df.select([
        pl.col(col).null_count().alias(f'{col}_missing') for col in df.columns
    ]).transpose(include_header=True, header_name='column', column_names=['missing_count'])

    missing_df = missing_analysis.with_columns([
        (pl.col('missing_count') / total_rows * 100).alias('missing_percent')
    ]).sort('missing_percent', descending=True)

    missing_with_nulls = missing_df.filter(pl.col('missing_count') > 0)
    if len(missing_with_nulls) == 0:
        return "No more missing values!"

    print(missing_with_nulls)

    missing_data = {
        'column': missing_with_nulls['column'].to_list(),
        'missing_percent': missing_with_nulls['missing_percent'].to_list()
    }
    fig = px.bar(
        missing_data,
        x='column',
        y='missing_percent',
        title='Missing values percent by column',
        labels={'missing_percent': 'Missing values percent', 'column': 'columns'},
        color='missing_percent',
        color_continuous_scale='Reds'
    )
    fig.update_layout(
        xaxis_tickangle=-45,
        height=500,
        showlegend=False
    )
    fig.show()

analyze_missing_values_by_column(df_pl)
```

2. `living_area`, `kitchen_area` - заполним на основе корреляции с `total_area`

```python
correlation = df_processed.select([
    pl.corr('total_area', 'living_area').alias('correlation')
])['correlation'][0]
print(f'Корреляция между total_area и living_area: {correlation:.3f}')

stats_df = df_processed.filter(
    pl.col('living_area').is_not_null() & pl.col('total_area').is_not_null()
).select([
    pl.mean('total_area').alias('mean_x'),
    pl.mean('living_area').alias('mean_y'),
    pl.std('total_area').alias('std_x'),
    pl.std('living_area').alias('std_y'),
    pl.corr('total_area', 'living_area').alias('corr')
])

mean_x = stats_df['mean_x'][0]
mean_y = stats_df['mean_y'][0]
std_x = stats_df['std_x'][0]
std_y = stats_df['std_y'][0]
corr = stats_df['corr'][0]

slope = corr * (std_y / std_x)
intercept = mean_y - slope * mean_x

df_processed = df_processed.with_columns(
    pl.when(pl.col('living_area').is_null())
    .then(slope * pl.col('total_area') + intercept)
    .otherwise(pl.col('living_area'))
    .alias('living_area')
)

correlation_k = df_processed.select([
    pl.corr('total_area', 'kitchen_area').alias('correlation')
])['correlation'][0]
print(f'Корреляция между total_area и kitchen_area: {correlation_k:.3f}')

stats_kitchen = df_processed.filter(
    pl.col('kitchen_area').is_not_null() & pl.col('total_area').is_not_null()
).select([
    pl.mean('total_area').alias('mean_x'),
    pl.mean('kitchen_area').alias('mean_y'),
    pl.std('total_area').alias('std_x'),
    pl.std('kitchen_area').alias('std_y'),
    pl.corr('total_area', 'kitchen_area').alias('corr')
])

mean_x_k = stats_kitchen['mean_x'][0]
mean_y_k = stats_kitchen['mean_y'][0]
std_x_k = stats_kitchen['std_x'][0]
std_y_k = stats_kitchen['std_y'][0]
corr_k = stats_kitchen['corr'][0]

slope_k = corr_k * (std_y_k / std_x_k)
intercept_k = mean_y_k - slope_k * mean_x_k

df_processed = df_processed.with_columns(
    pl.when(pl.col('kitchen_area').is_null())
    .then(slope_k * pl.col('total_area') + intercept_k)
    .otherwise(pl.col('kitchen_area'))
    .alias('kitchen_area')
)
```

4. `balcony` - заполним пропущенные значения с помощью knn на основе общей площади и количества комнат.

```python
from sklearn.neighbors import KNeighborsClassifier
from sklearn.preprocessing import StandardScaler

def fill_balcony_with_knn(df_processed, feature_columns):
    df = df_processed.clone()
    train_df = df.filter(pl.col('balcony').is_not_null())
    predict_df = df.filter(pl.col('balcony').is_null())

    X_train = train_df.select(feature_columns).to_numpy()
    y_train = train_df['balcony'].to_numpy()
    X_pred = predict_df.select(feature_columns).to_numpy()

    scaler = StandardScaler()
    X_train_scaled = scaler.fit_transform(X_train)
    X_pred_scaled = scaler.transform(X_pred)

    knn = KNeighborsClassifier(n_neighbors=min(5, len(X_train)), weights='distance')
    knn.fit(X_train_scaled, y_train)
    predictions = knn.predict(X_pred_scaled)

    null_indices = df.select(pl.arg_where(pl.col('balcony').is_null())).to_series()
    balcony_values = df['balcony'].to_list()
    for idx, prediction in zip(null_indices, predictions):
        balcony_values[idx] = prediction

    return df.with_columns(balcony=pl.Series(balcony_values))


print(f'Shape before: {df_processed.shape}')
print(f'Null values in balcony before: {df_processed["balcony"].null_count()}')
df_processed = fill_balcony_with_knn(df_processed, feature_columns=['total_area', 'rooms'])
print(f'Shape after: {df_processed.shape}')
print(f'Null values in balcony after: {df_processed["balcony"].null_count()}')
```

Сначала я подумал еще о том, что можно все пропуски в этих колонках выставить в какое-то константное большое значение (например, в удвоенный максимум из известных значений), и вместе с этим добавить булевы признаки `does_nearest_park_exist` или что-то в этом роде, но в таком случае нужно бы пойти к аналитикам и проконсультироваться, а является ли эта фича в целом обязательной при заполнении объявления и нужно ли "штрафовать" таким образом объявления, в которых эти значения не указаны.

```python
for column in ['parks_nearest', 'ponds_nearest', 'airports_distance', 'centers_distance']:
    existing_values = df_processed[column].drop_nulls().to_list()
    missing_count = df_processed[column].null_count()

    if missing_count > 0:
        random_values = np.random.choice(existing_values, size=missing_count, replace=True)
        null_indices = df_processed.select(
            pl.int_range(0, pl.len()).filter(pl.col(column).is_null())
        ).to_series().to_list()

        new_column = df_processed[column].to_list()
        for idx, replacement in zip(null_indices, random_values):
            new_column[idx] = replacement

        df_processed = df_processed.with_columns(pl.Series(column, new_column))
```

Сравним распределения `parks_nearest` до и после заполнения пропусков случайными известными значениями.

```python
from scipy.stats import gaussian_kde
import numpy as np

def plot_distribution_comparison(df_new, df_old, column):
    data1 = df_new.filter(pl.col(column).is_not_null())[column]
    kde1 = gaussian_kde(data1)
    x_range1 = np.linspace(data1.min(), data1.max(), 100)
    y_range1 = kde1(x_range1)

    data2 = df_old.filter(pl.col(column).is_not_null())[column]
    kde2 = gaussian_kde(data2)
    x_range2 = np.linspace(data2.min(), data2.max(), 100)
    y_range2 = kde2(x_range2)

    fig = go.Figure()
    fig.add_trace(go.Scatter(x=x_range1, y=y_range1, mode='lines',
                            name='KDE (new)', line=dict(color='blue')))
    fig.add_trace(go.Scatter(x=x_range2, y=y_range2, mode='lines',
                            name='KDE (old)', line=dict(color='red')))

    fig.add_trace(go.Histogram(
        x=data1,
        histnorm='probability density',
        opacity=0.3,
        name='histogram (new)',
        marker_color='blue'
    ))
    fig.add_trace(go.Histogram(
        x=data2,
        histnorm='probability density',
        opacity=0.3,
        name='histogram(old)',
        marker_color='red'
    ))

    fig.update_layout(
        xaxis_title=f'{column} values',
        yaxis_title='probability density',
        barmode='overlay'
    )
    fig.update_traces(opacity=0.6)
    fig.show()

plot_distribution_comparison(df_processed, df_pl, 'parks_nearest')
```

А вот, что получилось бы, если бы мы заменили пропуски в `parks_nearest` и других расстояниях медианным значением:

```python
parks_nearest_median = df_pl['parks_nearest'].median()
df_badly_processed = df_pl.with_columns(
    pl.col('parks_nearest').fill_null(parks_nearest_median)
)
plot_distribution_comparison(df_badly_processed, df_pl, 'parks_nearest')
```

Очевидно, что полных дубликатов в семплах уже не будет, так как после удаления или заполнения пропусков в каких-то фичах я таким образом мог сам себя обмануть и добавить зашумленность в данные (например, в стратегии заполнения случайными известными значениям).

```python
def check_lognormality(data, alpha=0.05):
    data_positive = data[data > 0]

    shape, loc, scale = stats.lognorm.fit(data_positive, floc=0)
    ks_stat, ks_p = stats.kstest(data_positive, 'lognorm', args=(shape, loc, scale))

    print(f'Kolmogorov-Smirnov test p-value: {ks_p:.4f}')
    print(f'Lognormal parameters - shape: {shape:.4f}, scale: {scale:.4f}')
    log_data = np.log(data)
    skewness = stats.skew(log_data)
    kurtosis = stats.kurtosis(log_data)
    print(f'Skewness: {skewness:.4f}')
    print(f'Kurtosis: {kurtosis:.4f}')
    return ks_p > alpha
```

Взглянем также на распределение каждого признака и соответсвующее ему нормальное. То есть при логнормальном распределении признака достаточно будет логарифмировать.

## 3. Создание новых признаков (Feature Engineering)


1. `living_area_ratio`
Проще говоря, это показатель того, насколько "жилой" получилась квартира. Если почти вся площадь - это комнаты, где можно жить (а не коридоры и кладовки), то значение будет высоким. Такие квартиры обычно (по логике) ценятся больше.

2. `log_area_per_room`
Здесь мы смотрим, насколько просторные в квартире комнаты. Делим общую площадь на число комнат - получаем "площадь на комнату". А логарифм нужен чтобы огромные квартиры не перекосили статистику.

3. `floor_ratio`
Эта фича показывает, на каком этаже находится квартира относительно всего дома. Например, "второй этаж из пяти" - это 0.4, а "второй этаж из десяти" - уже 0.2. Так мы понимаем, высокий это этаж или низкий в контексте конкретного дома. Избавляемся от дискретных фич этажности и этажа квартиры.

4. `log_living_efficiency`
Это "умная версия" первой фичи. Мы смотрим, какая часть полезной площади (всё кроме кухни) отведена под жилые комнаты. Если после вычета кухни остались в основном жилые помещения, то планировка эффективная. Если много коридоров и прихожих - не очень.

5. `log_area_height`
Мы считаем не просто площадь, а "объём" квартиры. Умножаем площадь на высоту потолков - получаем условный объём. Квартиры с высокими потолками кажутся просторнее, даже при одинаковой площади. Логарифм снова помогает не переоценивать экстремальные случаи.

```python
df_features = df_cleared.clone()

# Кажется, тут все говорит само за себя
df_features = df_features.with_columns([
    (pl.col('living_area') / pl.col('total_area')).alias('living_area_ratio'),
    (pl.col('total_area') / pl.col('rooms')).log1p().alias('log_area_per_room'),
    (pl.col('floor') / pl.col('floors_total')).alias('floor_ratio'),
    (pl.col('living_area') / (pl.col('total_area') - pl.col('kitchen_area'))).log1p().alias('log_living_efficiency'),
    (pl.col('total_area') * pl.col('ceiling_height')).log1p().alias('log_area_height'),
])

# Логарифмирую все логнормальные распределения (их графики в анализе выше в обработке аномалий)
key_log_features = []
for category in key_numeric_features:
    new_column = f'log_{category}'
    key_log_features.append(new_column)
    df_features = df_features.with_columns([
        pl.col(category).log1p().alias(new_column),
    ])

print(f'Quantity of the features after feature engineering: {df_features.width}')
print(f'Quantity of new features: {df_features.width - df_cleared.width}')


advanced_features = [
    'living_area_ratio',
    'log_area_per_room',
    'floor_ratio',
    'log_living_efficiency',
    'log_area_height',
]

fig_advanced = make_subplots(
    rows=2, cols=3,
    subplot_titles=[str(feature) for feature in advanced_features],
    vertical_spacing=0.15,
    horizontal_spacing=0.08
)

for i, feature in enumerate(advanced_features):
    if feature in df_features.columns:
        row = i // 3 + 1
        col_pos = i % 3 + 1

        feature_data = df_features[feature].to_numpy()

        fig_advanced.add_trace(
            go.Histogram(
                x=feature_data,
                name=feature,
                showlegend=False,
                nbinsx=40,
                opacity=0.8,
            ),
            row=row, col=col_pos
        )

fig_advanced.update_layout(
    height=800,
    showlegend=False,
    font=dict(size=12)
)

fig_advanced.show()
```

Будем пытаться определить оптимальное число кластеров методом локтя

```python
from sklearn.metrics import silhouette_score
from sklearn.cluster import KMeans

inertia = []
silhouette_scores = []
k_range = range(2, 11)

for k in k_range:
    kmeans = KMeans(n_clusters=k, verbose=0, random_state=SEED)
    cluster_labels = kmeans.fit_predict(X_cluster)
    inertia.append(kmeans.inertia_)
    silhouette_avg = silhouette_score(X_cluster, cluster_labels)
    silhouette_scores.append(silhouette_avg)


fig = make_subplots(rows=1, cols=2, subplot_titles=('elbow method', 'silhouette analysis'))

fig.add_trace(
    go.Scatter(x=list(k_range), y=inertia, mode='lines+markers'),
    row=1, col=1
)

fig.add_trace(
    go.Scatter(x=list(k_range), y=silhouette_scores, mode='lines+markers'),
    row=1, col=2
)

fig.update_layout(
    height=500,
    width=1000
)

fig.update_xaxes(title_text='k_clusters', row=1, col=1)
fig.update_xaxes(title_text='k_clusters', row=1, col=2)
fig.update_yaxes(title_text='Inertia', row=1, col=1)
fig.update_yaxes(title_text='Silhouette score', row=1, col=2)

fig.show()
```

Видимо, пока остановимся на 8 кластерах. Не вижу выраженного значения по методу локтя (возможно, беру недостаточное количество кластеров). В анализе силуэтов ожидал увидеть немного другой график и хотел фиксировать точку локального максимума вблизи 8, но, видимо, не сложилось.

Здесь пояснения давать точно не возьмусь, так как в этом задании проще использовать заключения и выводы по количеству кластеров как вывод black-box метода. (хотя ничего криминального, просто прогнали kmeans на нескольких значениях гиперпараметра `n_clusters` и проанализировали облака точек).

## 7. Построение и обучение модели

Неужели мы дошли?

**Цель:**
разработать модель регрессии, которая с достаточной точностью оценивает  логарифм цены квартиры `log_price` по набору признаков, описывающих её свойства и местоположение.

```python
df_reg_pd = df_normed.clone().to_pandas()

target = 'log_price'
features = [col for col in df_reg_pd.columns if col != target]

X, y = df_reg_pd[features], df_reg_pd[target]
```

Сейчас в данных уже нет дисбаланса классов для категориальных переменных, поэтому про стратификацию забываем, сплитим данные на тренировочную, тестовую и валидационную подвыборки.

Тестовая подвыборка будет составлять 10% от всех сэмплов.
Валидационная - 22% от оставшихся 90% сэмплов.

```python
from sklearn.model_selection import train_test_split

X_temp, X_test, y_temp, y_test = train_test_split(X, y, test_size=0.1, random_state=SEED)
X_train, X_val, y_train, y_val = train_test_split(X_temp, y_temp, test_size=0.22, random_state=SEED)

print(f'Train subset: {X_train.shape[0]} ({X_train.shape[0]/X.shape[0]:.2%})')
print(f'Val subset: {X_val.shape[0]} ({X_val.shape[0]/X.shape[0]:.2%})')
print(f'Test subset: {X_test.shape[0]} ({X_test.shape[0]/X.shape[0]:.2%})')
```

По набору метрик все банально: буду использовать стандратные метрики в задаче регрессии.

Не забываем о том, что мы говорим все же о логарифмированной цене (в качестве таргета), поэтому для подсчета метрики MAPE потенциируем оба (предсказанное и реальное) значения (это будет нашей доменной метрикой)

```python
from sklearn.metrics import r2_score, mean_squared_error, mean_absolute_error
import numpy as np

def evaluate_model(model, X, y):
    y_pred = model.predict(X)

    r2 = r2_score(y, y_pred)
    rmse = np.sqrt(mean_squared_error(y, y_pred))
    mae = mean_absolute_error(y, y_pred)

    y_true_exp = np.expm1(y)
    y_pred_exp = np.expm1(y_pred)
    mape = np.mean(np.abs((y_true_exp - y_pred_exp) / y_true_exp)) * 100

    return {'r2': r2, 'rmse': rmse, 'mae': mae, 'mape': mape}
```

Обучаем модели и оцениваем на валидационной подвыборке

```python
from sklearn.linear_model import LinearRegression, Ridge, Lasso, ElasticNet
from sklearn.ensemble import RandomForestRegressor, GradientBoostingRegressor
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import RobustScaler
import xgboost as xgb

models = {
    'Linear Regression': LinearRegression(),
    'Ridge': Ridge(alpha=1.0, random_state=SEED),
    'Lasso': Lasso(alpha=0.01, random_state=SEED),
    'ElasticNet':ElasticNet(alpha=0.01, l1_ratio=0.5, random_state=SEED),
    'Random Forest': RandomForestRegressor(random_state=SEED),
    'Gradient Boosting': GradientBoostingRegressor(random_state=SEED),
    'XGBoost': xgb.XGBRegressor(n_estimators=100, learning_rate=0.1, random_state=SEED)
}

results = {}
for name, model in models.items():
    model.fit(X_train, y_train)
    results[name] = evaluate_model(model, X_val, y_val)
```

```python
results_df = pd.DataFrame(results).T
display(results_df)

def show_model_comparison(metric_label, results_df):
    fig = px.bar(results_df, y=results_df.index, x=metric_label,
                title=f'Сравнение моделей по {metric_label}',
                orientation='h',
                color=metric_label,
                color_continuous_scale='Reds_r')
    fig.update_layout(height=500, width=800)
    fig.show()

show_model_comparison('r2', results_df)
show_model_comparison('rmse', results_df)
show_model_comparison('mae', results_df)
show_model_comparison('mape', results_df)
```

Вот такие получились бейзлайны, формально задача выполнена.

Выбираем лучшую модель по доменной метрике

```python
best_model_name = results_df['mape'].idxmin()
best_model = models[best_model_name]
print(f'Лучшая модель по mape: {best_model_name}')

print('\nОценка лучшей модели на тестовой выборке:')
print(evaluate_model(best_model, X_test, y_test))


from sklearn.model_selection import cross_val_score

print('\nРезультаты 5-кратной кросс-валидации для лучшей модели:')
cv_scores = cross_val_score(best_model, X_train, y_train, cv=5, scoring='neg_mean_absolute_percentage_error')
print(f'mape (кросс-валидация): {-100 * cv_scores.mean():.4f} +- {100 * cv_scores.std():.4f}')
```

А вот здесь уже интересно.

Кастомная фича (`log_area_height`), созданная перемножением нормально коррелирующих признаков в importance сильно отличается от `log_total_area` и `log_ceiling_height`.

Визуализируем предсказания XGBoost на тестовой подвыборке.

```python
y_pred = best_model.predict(X_test)

pred_df = pd.DataFrame({'actual': y_test, 'predicted': y_pred, 'residual': y_test - y_pred})

fig = make_subplots(rows=1, cols=3,
                   subplot_titles=[f'Фактические vs Предсказанные значения ({best_model_name})',
                                 f'Остатки vs Предсказанные значения ({best_model_name})',
                                 f'Распределение остатков ({best_model_name})'])

fig.add_trace(
    go.Scatter(x=pred_df['actual'], y=pred_df['predicted'], mode='markers', name='predicted vs actual'),
    row=1, col=1
)
fig.add_trace(
    go.Scatter(x=[pred_df['actual'].min(), pred_df['actual'].max()],
               y=[pred_df['actual'].min(), pred_df['actual'].max()],
               mode='lines', name='ideal prediction',
               line=dict(color='red', dash='dash')),
    row=1, col=1
)

fig.add_trace(
    go.Scatter(x=pred_df['predicted'], y=pred_df['residual'], mode='markers', name='residuals'),
    row=1, col=2
)
fig.add_hline(y=0, line_dash='dash', line_color='red', row=1, col=2)


fig.add_trace(
    go.Histogram(x=pred_df['residual'], name='residual distribution'),
    row=1, col=3
)

fig.update_layout(height=500, width=1500, showlegend=False)
fig.show()
```
